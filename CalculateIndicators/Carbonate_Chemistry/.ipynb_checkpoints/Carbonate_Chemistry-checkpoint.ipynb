## Carbonate Chemistry

```python
import pandas as pn
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
import math
import xarray as xr
import scipy.interpolate as sio
```

```python
# This cell just defines the function that extracts the transect 1 data from a csv file.  The output is just a table/dataframe (just like in R!)
# with only the values for transect 1
def transect_1(csv_file):
    df = pn.read_csv(csv_file)
    
    # turn the numbered stations into floats
    number_station = np.empty(len(df.Station))
    for i in range(len(df)):
        if type(df.Station[i]) == str and (len(df.Station[i])>4):
            number_station[i] = np.nan
        else:
            number_station[i] = float(df.Station[i])
    
    # add a column for this new float
    df['station_number'] = number_station
    
    if np.any(np.unique(np.floor(df.station_number)) == 1):
        transect1 = df[np.floor(df.station_number) == 1]
    else:
        print('No transect 1')
            
    distance1 = np.empty(len(np.where((np.floor(np.unique(df.station_number))==1))[0]))
    distance1[:] = np.nan

    R = 6373.0 #radius of the Earth
    D = 0
    stations1 = np.unique(transect1.station_number)
    
    for i in range(len(stations1)-1):
        station_subset = transect1[transect1.station_number == stations1[i]]
        station_subset_next = transect1[transect1.station_number == stations1[i+1]]
    
        lat1 = np.radians(np.unique(station_subset.Lat)[0]) #coordinates
        lon1 = np.radians(np.unique(station_subset.Lon)[0])

        lat2 = np.radians(np.unique(station_subset_next.Lat)[0])
        lon2 = np.radians(np.unique(station_subset_next.Lon)[0])

        dlon = lon2 - lon1 #change in coordinates
        dlat = lat2 - lat1

        a = math.sin(dlat / 2)**2 + math.cos(lat1) * math.cos(lat2) * math.sin(dlon / 2)**2 #Haversine formula
        c = 2 * math.atan2(math.sqrt(a), math.sqrt(1 - a))

        distance1[i] = (R * c)
        
    transect1['along_track_distance'] = np.nan
    transect1.along_track_distance[transect1.station_number == stations1[0]] = 0
    for i in range(len(stations1)-1):
        transect1.along_track_distance[transect1.station_number == stations1[i+1]] = np.sum(distance1[:i+1])

    return (transect1)   
```

```python
# This cell just defines the function that extracts the transect 6 data from a csv file.  The output is just a table/dataframe (just like in R!)
# with only the values for transect 6
def transect_6(csv_file):
    df = pn.read_csv(csv_file)
    
    # turn the numbered stations into floats
    number_station = np.empty(len(df.Station))
    for i in range(len(df)):
        if type(df.Station[i]) == str and (len(df.Station[i])>4):
            number_station[i] = np.nan
        else:
            number_station[i] = float(df.Station[i])
    
    # add a column for this new float
    df['station_number'] = number_station
    
    if np.any(np.unique(np.floor(df.station_number)) ==6):
        transect6 = df[np.floor(df.station_number) == 6]
    else:
        print('No transect 6')
            
    
    distance6 = np.empty(len(np.where((np.floor(np.unique(df.station_number))==6))[0]))
    distance6[:] = np.nan

    R = 6373.0 #radius of the Earth
    D = 0

    stations6 = np.unique(transect6.station_number)
    
    for i in range(len(stations6)-1):
        station_subset = transect6[transect6.station_number == stations6[i]]
        station_subset_next = transect6[transect6.station_number == stations6[i+1]]
    
        lat1 = np.radians(np.unique(station_subset.Lat)[0]) #coordinates
        lon1 = np.radians(np.unique(station_subset.Lon)[0])

        lat2 = np.radians(np.unique(station_subset_next.Lat)[0])
        lon2 = np.radians(np.unique(station_subset_next.Lon)[0])

        dlon = lon2 - lon1 #change in coordinates
        dlat = lat2 - lat1

        a = math.sin(dlat / 2)**2 + math.cos(lat1) * math.cos(lat2) * math.sin(dlon / 2)**2 #Haversine formula
        c = 2 * math.atan2(math.sqrt(a), math.sqrt(1 - a))

        distance6[i] = (R * c)

        
    transect6['along_track_distance'] = np.nan
    transect6.along_track_distance[transect6.station_number == stations6[0]] = 0
    for i in range(len(stations6)-1):
        transect6.along_track_distance[transect6.station_number == stations6[i+1]] = np.sum(distance6[:i+1])
    return (transect6)   
```

```python
transect_1_lon = [-71.9,-71.89916667,-71.81083333,-71.72033333,-71.63483333,-71.54466667,-71.46083333]
transect_1_lat = [41.001,40.96066667,40.75866667,40.86946667,40.47173333,40.0445,39.80066667]
topo = xr.open_dataset("/Users/nyelab/Downloads/ETOPO1_Bed_g_gmt4(1).grd")
transect_1_seafloor_depth = np.empty(7)
transect_1_seafloor_depth[:] = np.nan

for i in range(len(transect_1_seafloor_depth)):

    lonvalue = transect_1_lon[i]
    latvalue = transect_1_lat[i]

    absolute_val_array_lon = np.abs(topo.x - lonvalue)
    absolute_val_array_lat = np.abs(topo.y - latvalue)

    smallest_difference_index_lon = absolute_val_array_lon.argmin()
    smallest_difference_index_lat = absolute_val_array_lat.argmin()

    transect_1_seafloor_depth[i] = topo.z[smallest_difference_index_lat,smallest_difference_index_lon]

distance1 = np.empty(6)
distance1[:] = np.nan

R = 6373.0 #radius of the Earth
D = 0

    
for i in range(6):
    
    lat1 = np.radians(transect_1_lat[i]) #coordinates
    lon1 = np.radians(transect_1_lon[i])

    lat2 = np.radians(transect_1_lat[i+1])
    lon2 = np.radians(transect_1_lon[i+1])

    dlon = lon2 - lon1 #change in coordinates
    dlat = lat2 - lat1

    a = math.sin(dlat / 2)**2 + math.cos(lat1) * math.cos(lat2) * math.sin(dlon / 2)**2 #Haversine formula
    c = 2 * math.atan2(math.sqrt(a), math.sqrt(1 - a))

    distance1[i] = (R * c)
i=0
distance_t1 = [0,np.sum(distance1[i:i+1])*1000,np.sum(distance1[i:i+2])*1000,
              np.sum(distance1[i:i+3])*1000,np.sum(distance1[i:i+4])*1000,
              np.sum(distance1[i:i+5])*1000,np.sum(distance1[i:i+6])*1000]

# linear interpolation of seafloor depth between ctd cast points
y2 = transect_1_seafloor_depth
x2 = distance_t1
xq = np.linspace(0,distance_t1[6], 17)
y_new = np.interp(xq,x2,y2)

depth = y_new
depth[depth<-100] = -100

D = np.diff(xq)
d = np.diff(depth)
area = np.empty(16)

for i in range(16):
    X = D[i]/1000
    Y = depth[i]/1000
    area_rect = X*Y
    Y2 = d[i]/1000
    area_triangle = 0.5 * X*Y2
    area[i] = (area_rect*(-1)) + (area_triangle*(-1))
area

transect1_area = np.sum(area)
```

```python
# This defines a function that allows you to pick a variable from your transect data frame, interpolate the data from the different CTD casts,
# plots the interpolated data on a graph with the contour level of your choosing highlighted and with a title that you pick
# It also spits out the area within that contour level and the mean value of the variable within that contour level

def area_within_contour(transect_data, variable, contour_level, title_of_plot, T1_or_T6):
    # Next establish the seafloor depth
    topo = xr.open_dataset("/Users/nyelab/Downloads/ETOPO1_Bed_g_gmt4(1).grd")
    transect = transect_data #fill in the name of the transect

    #  Find the depth at each ctd cast site
    lons = np.unique(transect['Lon'])
    lats = np.flip(np.unique(transect['Lat']))
    seafloor_depth = np.empty(len(np.unique(transect['Lat'])))
    seafloor_depth[:] = np.nan
    for i in range(len(seafloor_depth)):

        lonvalue = lons[i]
        latvalue = lats[i]

        absolute_val_array_lon = np.abs(topo.x - lonvalue)
        absolute_val_array_lat = np.abs(topo.y - latvalue)

        smallest_difference_index_lon = absolute_val_array_lon.argmin()
        smallest_difference_index_lat = absolute_val_array_lat.argmin()

        seafloor_depth[i] = topo.z[smallest_difference_index_lat,smallest_difference_index_lon]


    print('seafloor depths,', seafloor_depth)
    
    # Preforma a gridded interpolation of whatever variable you are looking at
    y= np.arange(0,110,10)
    x= np.arange(0,240000, 10000)
    X, Y = np.meshgrid(x,y)
    grid_transect = sio.griddata((transect['along_track_distance']*1000, transect['Depth [m]']), transect[variable],(X,Y))
    
    # linear interpolation of seafloor depth between ctd cast points
    y2 = seafloor_depth
    x2 = np.unique(transect['along_track_distance']*1000)
    xq = x
    y_new = np.interp(xq,x2,y2)
    
    # delete all the interpolated stuff below the seafloor depth
    # Find the depth at each ctd cast site
    delete_index = np.empty(len(xq))
    for i in range(len(xq)):

        value = abs(y_new[i])

        absolute_val_array = np.abs(y - value)

        smallest_difference_index = absolute_val_array.argmin()

        delete_index[i] = smallest_difference_index +1
        
    for i in range(len(xq)):
        num = int(delete_index[i])
        grid_transect[num:,i] = np.nan  
        
    # Plot the data
    fig, ax = plt.subplots(figsize = [25,10])
    c = plt.contourf(x,y,grid_transect, levels = 15, cmap = 'rainbow')
    c2 = plt.contour(x,y,grid_transect,levels = [contour_level])
    plt.ylim([100,0])
    plt.colorbar(c)
    plt.clabel(c2)
    plt.scatter(transect['along_track_distance']*1000, transect['Depth [m]'],c = 'k')
    ax.tick_params(labelsize = 16)
    ax.set_ylabel('Depth (m)', fontsize = 16)
    ax.set_xlabel('Along Section Distance (m)', fontsize = 16)
    ax.set_title(title_of_plot, fontsize = 18)
    
    area_within_contour = (len(grid_transect[grid_transect<contour_level])*10*10000)/(1000*1000)
    average = np.nanmean(grid_transect[grid_transect<contour_level])
    
    if T1_or_T6 == 'T1':
        percent_area = (area_within_contour/transect1_area)*100
    if T1_or_T6 == 'T6':
        percent_area = (area_within_contour/transect6_area)*100
        
    print('area within contour is ', area_within_contour, ' km')
    print('percent transect area within contour is ', percent_area, '%')
    print('average value within contour is ', average)
    return(area_within_contour)
```

```python
distance = np.empty(4)
distance[:] = np.nan
import math
R = 6373.0 #radius of the Earth
D = 0
stations = np.unique(transect6.station_number)
for i in range(4):
    station_subset = transect6[transect6.station_number == stations[i]]
    station_subset_next = transect6[transect6.station_number == stations[i+1]]
    
    lat1 = np.radians(np.unique(station_subset.Latitude)[0]) #coordinates

    lon1 = np.radians(np.unique(station_subset.Longitude)[0])

    lat2 = np.radians(np.unique(station_subset_next.Latitude)[0])

    lon2 = np.radians(np.unique(station_subset_next.Longitude)[0])

    dlon = lon2 - lon1 #change in coordinates

    dlat = lat2 - lat1


    a = math.sin(dlat / 2)**2 + math.cos(lat1) * math.cos(lat2) * math.sin(dlon / 2)**2 #Haversine formula

    c = 2 * math.atan2(math.sqrt(a), math.sqrt(1 - a))

    distance[i] = (R * c)
```

```python
def inNYB_nodepth(longitude, latitude, var, shape):
    X = longitude
    Y = latitude
    x, y = np.meshgrid(X, Y)
    empty = np.empty([len(latitude), len(longitude)])
    empty[:,:] = np.nan
    for i in range(len(latitude)):
        for j in range(len(longitude)):
            if Point(x[i,j], y[i,j]).within(shape) == True:
                empty[i,j] = var[i,j]
    return empty
```

```python
def percent_WAr_ColdPool(filepath, omega_val, topo, remove_stns, depthcutoff, nyb_vol):
    data = pn.read_csv(filepath)
    Stns = np.unique(data.Station) # get unique stations
    if np.any(remove_stns):
        for i in remove_stns:
            Stns = Stns[Stns!=i]
            
    # RUN THIS
    WAr_depth = np.empty(len(Stns))
    WAr_depth[:] = np.nan
    T_depth = np.empty(len(Stns))
    T_depth[:] = np.nan
    S_depth = np.empty(len(Stns))
    S_depth[:] = np.nan
    Lat_WAr =np.empty(len(Stns))
    Lat_WAr[:] = np.nan
    Lon_WAr = np.empty(len(Stns))
    Lon_WAr[:] = np.nan
    
    for i in range(len(np.unique(Stns))):
        subset = data[data.Station == Stns[i]]
        if len(subset) >1:
            S_func = sio.interp1d(subset['Salinity'], subset['Depth [m]'], bounds_error = False)
            T_func = sio.interp1d(subset['pTemp [°C]'], subset['Depth [m]'], bounds_error = False)
            WAr_func = sio.interp1d(subset['WAr out'], subset['Depth [m]'], bounds_error = False)
            output1 = WAr_func(omega_val)
            output2 = T_func(10)
            output3 = S_func(34)
            if np.isnan(output1) == True:
                WAr_depth[i] = np.unique(subset['Bottom Depth [m]'])
            if np.isnan(output1) == False:
                WAr_depth[i] = output1
            if np.isnan(output2) == True:
                T_depth[i] = np.unique(subset['Bottom Depth [m]'])
            if np.isnan(output2) == False:
                T_depth[i] = output2
            if np.isnan(output3) == True:
                if np.max(subset['Salinity'] < 34):
                    S_depth[i] = 0
                if np.max(subset['Salinity']> 34):
                    S_depth[i] = np.unique(subset['Bottom Depth [m]'])
            if np.isnan(output3) == False:
                S_depth[i] = output3
                
            Lat_WAr[i] = np.nanmean(subset.Lat)
            Lon_WAr[i] = np.nanmean(subset.Lon)
        
    
    
    # interpolate
    from scipy.interpolate import LinearNDInterpolator

    X = topo.x[6390:6515].values
    Y = topo.y[7710:7862]
    X, Y = np.meshgrid(X, Y)  # 2D grid for interpolation
    interp1 = LinearNDInterpolator(list(zip(Lon_WAr, Lat_WAr)), WAr_depth)
    Z_war = interp1(X, Y)
    p = plt.pcolormesh(X, Y, Z_war, shading='auto', vmax = 50, vmin = 0)
    plt.scatter(Lon_WAr, Lat_WAr, label="input point")
    plt.legend()
    plt.colorbar(p)
    plt.axis("equal")
    plt.title('WAr', fontsize = 14)
    plt.show()
    
    interp2 = LinearNDInterpolator(list(zip(Lon_WAr, Lat_WAr)), T_depth)
    Z_t = interp2(X, Y)
    p2 = plt.pcolormesh(X, Y, Z_t, shading='auto', vmax = 50, vmin = 0)
    plt.scatter(Lon_WAr, Lat_WAr,label="input point")
    plt.legend()
    plt.colorbar(p2)
    plt.axis("equal")
    plt.title('Temp', fontsize = 14)
    plt.show()
    
    interp3 = LinearNDInterpolator(list(zip(Lon_WAr, Lat_WAr)), S_depth)
    Z_s = interp3(X, Y)
    p3 = plt.pcolormesh(X, Y, Z_s, shading='auto', vmax = 50, vmin =0)
    plt.scatter(Lon_WAr, Lat_WAr,label="input point")
    plt.legend()
    plt.colorbar(p3)
    plt.axis("equal")
    plt.title("Salinity", fontsize = 14)
    plt.show()
    
    # make a specific thickness matrix
    z_war = (-1*topo.z[7710:7862,6390:6515]) - Z_war 
    
    #account for places with topographical mismatch.  
    #There are a few places where the topography was shallower than the interpolated thickness.  In these places we will make the thickness 0

    #if the value is greater than 200 we will put 200, we are only considering the upper 200 m

    thickness_war = np.empty(np.shape(z_war))
    thickness_war[:,:] = np.nan
    for i in range(np.shape(z_war)[0]):
        for j in range(np.shape(z_war)[1]):
            if z_war[i,j] < 0: #this would indicate that the topography is shallower than the interpolated depth
                if -1*topo.z[7710+i,6390+j]>depthcutoff: # if the topography is deeper than the depth cut off
                    thickness_war[i,j] = depthcutoff
                else:
                    thickness_war[i,j] = -1*topo.z[7710+i,6390+j] # if the topography is shallower than the depthcutoff just use the topography
            else:
                if z_war[i,j] < depthcutoff: #if thickness of 1.5 is shallower than the depth cutoff
                    thickness_war[i,j] = z_war[i,j] 
                if z_war[i,j] > depthcutoff: # if the thickness is greater than the depth cutoff
                    if depthcutoff - Z_war[i,j] < 0: # if the depthcutoff is shallower than the interpolated depth
                        thickness_war[i,j] = 0
                    if depthcutoff - Z_war[i,j] > 0:
                        thickness_war[i,j] = depthcutoff - Z_war[i,j]
    
    # thickness in terms of volume per little cube
    latitude = topo.y[7710:7862]
    longitude = topo.x[6390:6515]
    vol_thickness_war = np.empty(np.shape(thickness_war))
    vol_thickness_war[:,:] = np.nan
    for i in range(np.shape(z_war)[0]):
        for j in range(np.shape(z_war)[1]):
            vol_thickness_war[i,j] = thickness_war[i,j]*0.001*((np.cos(np.radians(latitude[j]))*111.321)*(0.01666667))*((111.200*0.01666667))
        
    volume_war = np.nansum(vol_thickness_war) # total volume is the sum of all of these little cubes in km3
    total_vol_war = np.empty(np.shape(vol_thickness_war))
    total_vol_war[:,:] = np.nan

    for i in range(np.shape(z_war)[0]):
        for j in range(np.shape(z_war)[1]):
            if np.isnan(z_war[i,j]) == False:
                if topo.z[7710+i,6390+j] < -depthcutoff:
                    total_vol_war[i,j] = -depthcutoff*0.001*((np.cos(np.radians(latitude[j]))*111.321)*(0.01666667))*((111.200*0.01666667))
                if topo.z[7710+i,6390+j] >= -depthcutoff:
                    total_vol_war[i,j] = topo.z[7710+i,6390+j]*0.001*((np.cos(np.radians(latitude[j]))*111.321)*(0.01666667))*((111.200*0.01666667))
        
    total_volume = -1*np.nansum(total_vol_war)
    
    Z_cp = np.empty(np.shape(Z_t))
    Z_cp[:,:]=np.nan

    for i in range(np.shape(z_war)[0]):
        for j in range(np.shape(z_war)[1]):
            if np.isnan(Z_t[i,j]) == False:
                if Z_t[i,j]>0 and Z_s[i,j] > 0:
                    if Z_t[i,j] > Z_s[i,j]:
                        Z_cp[i,j] = Z_t[i,j]
                    if Z_t[i,j] < Z_s[i,j]:
                        Z_cp[i,j] = Z_s[i,j]
                    if Z_t[i,j] == Z_s[i,j]:
                        Z_cp[i,j] = Z_t[i,j]
                else:
                    Z_cp[i,j] = 0
    
    z_cp = (-1*topo.z[7710:7862,6390:6515]) - Z_cp
    
    Z_war_cp = np.empty(np.shape(Z_t))
    Z_war_cp[:,:]=np.nan

    for i in range(np.shape(z_war)[0]):
        for j in range(np.shape(z_war)[1]):
            if np.isnan(Z_t[i,j]) == False:
                if Z_t[i,j]>0 and Z_s[i,j] > 0:
                    if Z_war[i,j] >0:
                        vector = [Z_t[i,j], Z_s[i,j], Z_war[i,j]]
                        deepest = np.max(vector)
                        if deepest == Z_t[i,j]:
                            Z_war_cp[i,j] = Z_t[i,j]
                        if deepest == Z_s[i,j]:
                            Z_war_cp[i,j] = Z_s[i,j]
                        if deepest == Z_war[i,j]:
                            Z_war_cp[i,j] = Z_war[i,j]
                    else:
                        Z_war_cp[i,j] = 0
                else:
                    Z_war_cp[i,j] = 0
    
    z_war_cp = (-1*topo.z[7710:7862,6390:6515]) - Z_war_cp
    
    c = plt.contourf(thickness_war, levels = np.arange(0,210,10))
    #plt.scatter(101,76)
    plt.colorbar(c)
    plt.title('WAr Thickness', fontsize = 14)
    plt.show()

    
    #account for places with topographical mismatch.  
    #There are a few places where the topography was shallower than the interpolated thickness.  In these places we will make the thickness 0

    #if the value is greater than 200 we will put 200, we are only considering the upper 200 m

    thickness_cp = np.empty(np.shape(z_cp))
    thickness_cp[:,:] = np.nan
    for i in range(np.shape(z_war)[0]):
        for j in range(np.shape(z_war)[1]):
            if z_cp[i,j] < 0:
                if -1*topo.z[7710+i,6390+j]>depthcutoff:
                    thickness_cp[i,j] = depthcutoff
                else:
                    thickness_cp[i,j] = -1*topo.z[7710+i,6390+j]
            else:
                if z_cp[i,j] < depthcutoff:
                    thickness_cp[i,j] = z_cp[i,j]
                if z_cp[i,j] > depthcutoff:
                    if depthcutoff - Z_cp[i,j] < 0:
                        thickness_cp[i,j] = 0
                    if depthcutoff - Z_cp[i,j] > 0:
                        thickness_cp[i,j] = depthcutoff - Z_cp[i,j]
                        
    c = plt.contourf(thickness_cp, levels = np.arange(0,210,10))
    #plt.scatter(101,76)
    plt.colorbar(c)
    plt.title('Cold Pool Thickness', fontsize = 14)
    plt.show()
    
    # thickness in terms of volume per little cube
    
    latitude = topo.y[7710:7862]
    longitude = topo.x[6390:6515]
    vol_thickness_cp = np.empty(np.shape(thickness_cp))
    vol_thickness_cp[:,:] = np.nan
    for i in range(np.shape(z_war)[0]):
        for j in range(np.shape(z_war)[1]):
            vol_thickness_cp[i,j] = thickness_cp[i,j]*0.001*((np.cos(np.radians(latitude[j]))*111.321)*(0.01666667))*((111.200*0.01666667))
    
    volume_cp = np.nansum(vol_thickness_cp) # total volume is the sum of all of these little cubes in km3

    thickness_war_cp = np.empty(np.shape(z_war_cp))
    thickness_war_cp[:,:] = np.nan
    for i in range(np.shape(z_war_cp)[0]):
        for j in range(np.shape(z_war_cp)[1]):
            if z_war_cp[i,j] < 0:
                if -1*topo.z[7710+i,6390+j]>depthcutoff:
                    thickness_war_cp[i,j] = depthcutoff
                else:
                    thickness_war_cp[i,j] = -1*topo.z[7710+i,6390+j]
            else:
                if z_war_cp[i,j] < depthcutoff:
                    thickness_war_cp[i,j] = z_war_cp[i,j]
                if z_war_cp[i,j] > depthcutoff:
                    if depthcutoff - Z_war_cp[i,j] < 0:
                        thickness_war_cp[i,j] = 0
                    if depthcutoff - Z_war_cp[i,j] > 0:
                        thickness_war_cp[i,j] = depthcutoff - Z_war_cp[i,j]
                        
    c = plt.contourf(thickness_war_cp, levels = np.arange(0,210,10))
    #plt.scatter(101,76)
    plt.colorbar(c)
    plt.title('Omega Cold Pool Overlap Thickness', fontsize = 14)
    plt.show()
    
    # thickness in terms of volume per little cube
    
    latitude = topo.y[7710:7862]
    longitude = topo.x[6390:6515]
    vol_thickness_war_cp = np.empty(np.shape(thickness_war_cp))
    vol_thickness_war_cp[:,:] = np.nan
    for i in range(np.shape(z_war_cp)[0]):
        for j in range(np.shape(z_war_cp)[1]):
            vol_thickness_war_cp[i,j] = thickness_war_cp[i,j]*0.001*((np.cos(np.radians(latitude[j]))*111.321)*(0.01666667))*((111.200*0.01666667))
    
    volume_war_cp = np.nansum(vol_thickness_war_cp) # total volume is the sum of all of these little cubes in km3

                        
        
    print('Volume of water below ' + str(omega_val) + ' = ' + str((volume_war/total_volume)*100))
    print('Volume of water in Cold Pool = ' + str((volume_cp/total_volume)*100))
    print('Total Volume analyzed = ' + str(total_volume))
    pct_omega = (volume_war/total_volume)*100
    pct_cp = (volume_cp/total_volume)*100
    pct_omega_cp = (volume_war_cp/total_volume)*100
    pct_omega_nyb = (volume_war/nyb_vol)*100
    pct_cp_nyb = (volume_cp/nyb_vol)*100
    pct_omega_cp_nyb = (volume_war_cp/nyb_vol)*100
    pct_nyb = (total_volume/nyb_vol)*100
    
    #return thickness_cp, thickness_war, 
    return np.unique(data.CruiseID), omega_val, 10, 34, total_volume, pct_nyb, pct_omega, pct_cp, pct_omega_cp, nyb_vol, pct_omega_nyb, pct_cp_nyb, pct_omega_cp_nyb
```

```python
def percent_WAr_ColdPool2(filepath, omega_val, topo, remove_stns):
    data = pn.read_csv(filepath)
    Stns = np.unique(data.Station) # get unique stations
    if np.any(remove_stns):
        for i in remove_stns:
            Stns = Stns[Stns!=i]
    # RUN THIS
    WAr_depth = np.empty(len(Stns))
    WAr_depth[:] = np.nan
    T_depth = np.empty(len(Stns))
    T_depth[:] = np.nan
    S_depth = np.empty(len(Stns))
    S_depth[:] = np.nan
    Lat_WAr =np.empty(len(Stns))
    Lat_WAr[:] = np.nan
    Lon_WAr = np.empty(len(Stns))
    Lon_WAr[:] = np.nan
    
    for i in range(len(np.unique(Stns))):
        subset = data[data.Station == Stns[i]]
        if len(subset) >1:
            S_func = sio.interp1d(subset['Salinity'], subset['Depth [m]'], bounds_error = False)
            T_func = sio.interp1d(subset['pTemp [°C]'], subset['Depth [m]'], bounds_error = False)
            WAr_func = sio.interp1d(subset['WAr out'], subset['Depth [m]'], bounds_error = False)
            output1 = WAr_func(omega_val)
            output2 = T_func(10)
            output3 = S_func(34)
            if np.isnan(output1) == True:
                WAr_depth[i] = np.unique(subset['Bottom Depth [m]'])
            if np.isnan(output1) == False:
                WAr_depth[i] = output1
            if np.isnan(output2) == True:
                T_depth[i] = np.unique(subset['Bottom Depth [m]'])
            if np.isnan(output2) == False:
                T_depth[i] = output2
            if np.isnan(output3) == True:
                if np.max(subset['Salinity'] < 34):
                    S_depth[i] = 0
                if np.max(subset['Salinity']> 34):
                    S_depth[i] = np.unique(subset['Bottom Depth [m]'])
            if np.isnan(output3) == False:
                S_depth[i] = output3
                
            Lat_WAr[i] = np.nanmean(subset.Lat)
            Lon_WAr[i] = np.nanmean(subset.Lon)
        
    
    
    # interpolate
    from scipy.interpolate import LinearNDInterpolator

    X = topo.x[6390:6515].values
    Y = topo.y[7710:7862]
    X, Y = np.meshgrid(X, Y)  # 2D grid for interpolation
    interp1 = LinearNDInterpolator(list(zip(Lon_WAr, Lat_WAr)), WAr_depth)
    Z_war = interp1(X, Y)
    p = plt.pcolormesh(X, Y, Z_war, shading='auto', vmax = 50, vmin = 0)
    plt.scatter(Lon_WAr, Lat_WAr, label="input point")
    plt.legend()
    plt.colorbar(p)
    plt.axis("equal")
    plt.title('WAr', fontsize = 14)
    plt.show()
    
    interp2 = LinearNDInterpolator(list(zip(Lon_WAr, Lat_WAr)), T_depth)
    Z_t = interp2(X, Y)
    p2 = plt.pcolormesh(X, Y, Z_t, shading='auto', vmax = 50, vmin = 0)
    plt.scatter(Lon_WAr, Lat_WAr,label="input point")
    plt.legend()
    plt.colorbar(p2)
    plt.axis("equal")
    plt.title('Temp', fontsize = 14)
    plt.show()
    
    interp3 = LinearNDInterpolator(list(zip(Lon_WAr, Lat_WAr)), S_depth)
    Z_s = interp3(X, Y)
    p3 = plt.pcolormesh(X, Y, Z_s, shading='auto', vmax = 50, vmin =0)
    plt.scatter(Lon_WAr, Lat_WAr,label="input point")
    plt.legend()
    plt.colorbar(p3)
    plt.axis("equal")
    plt.title("Salinity", fontsize = 14)
    plt.show()
    
    for i in range(np.shape(Z_war[0])):
        for j in range(np.shape(Z_war[1])):
            if (-1*topo.z[7710+i,6390+j]) > 100:
                Z_war[i,j] = np.nan
                
    # make a specific thickness matrix
    z_war = (-1*topo.z[7710:7862,6390:6515]) - Z_war 
    
    #account for places with topographical mismatch.  
    #There are a few places where the topography was shallower than the interpolated thickness.  In these places we will make the thickness 0

    #if the value is greater than 200 we will put 200, we are only considering the upper 200 m

    thickness_war = np.empty(np.shape(z_war))
    thickness_war[:,:] = np.nan
    for i in range(np.shape(z_war)[0]):
        for j in range(np.shape(z_war)[1]):
            if z_war[i,j] < 0:
                if -1*topo.z[7710+i,6390+j]>100:
                    thickness_war[i,j] = np.nan
                else:
                    thickness_war[i,j] = -1*topo.z[7710+i,6390+j]
            else:
                if z_war[i,j] < 100:
                    thickness_war[i,j] = z_war[i,j]
                if z_war[i,j] > 100:
                    if 200 - Z_war[i,j] < 0:
                        thickness_war[i,j] = 0
                    if 200 - Z_war[i,j] > 0:
                        thickness_war[i,j] = 200 - Z_war[i,j]
    
    # thickness in terms of volume per little cube
    latitude = topo.y[7710:7862]
    longitude = topo.x[6390:6515]
    vol_thickness_war = np.empty(np.shape(thickness_war))
    vol_thickness_war[:,:] = np.nan
    for i in range(np.shape(z_war)[0]):
        for j in range(np.shape(z_war)[1]):
            vol_thickness_war[i,j] = thickness_war[i,j]*0.001*((np.cos(np.radians(latitude[j]))*111.321)*(0.01666667))*((111.200*0.01666667))
        
    volume_war = np.nansum(vol_thickness_war) # total volume is the sum of all of these little cubes in km3
    total_vol_war = np.empty(np.shape(vol_thickness_war))
    total_vol_war[:,:] = np.nan

    for i in range(np.shape(z_war)[0]):
        for j in range(np.shape(z_war)[1]):
            if np.isnan(z_war[i,j]) == False:
                if topo.z[7710+i,6390+j] < -200:
                    total_vol_war[i,j] = -200*0.001*((np.cos(np.radians(latitude[j]))*111.321)*(0.01666667))*((111.200*0.01666667))
                if topo.z[7710+i,6390+j] >= -200:
                    total_vol_war[i,j] = topo.z[7710+i,6390+j]*0.001*((np.cos(np.radians(latitude[j]))*111.321)*(0.01666667))*((111.200*0.01666667))
        
    total_volume = -1*np.nansum(total_vol_war)
    
    Z_cp = np.empty(np.shape(Z_t))
    Z_cp[:,:]=np.nan

    for i in range(np.shape(z_war)[0]):
        for j in range(np.shape(z_war)[1]):
            if np.isnan(Z_t[i,j]) == False:
                if Z_t[i,j]>0 and Z_s[i,j] > 0:
                    if Z_t[i,j] > Z_s[i,j]:
                        Z_cp[i,j] = Z_t[i,j]
                    if Z_t[i,j] < Z_s[i,j]:
                        Z_cp[i,j] = Z_s[i,j]
                    if Z_t[i,j] == Z_s[i,j]:
                        Z_cp[i,j] = Z_t[i,j]
                else:
                    Z_cp[i,j] = 0
    
    z_cp = (-1*topo.z[7710:7862,6390:6515]) - Z_cp
    
    c = plt.contourf(thickness_war, levels = np.arange(0,210,10))
    #plt.scatter(101,76)
    plt.colorbar(c)
    plt.title('WAr Thickness', fontsize = 14)
    plt.show()

    
    #account for places with topographical mismatch.  
    #There are a few places where the topography was shallower than the interpolated thickness.  In these places we will make the thickness 0

    #if the value is greater than 200 we will put 200, we are only considering the upper 200 m

    thickness_cp = np.empty(np.shape(z_cp))
    thickness_cp[:,:] = np.nan
    for i in range(np.shape(z_war)[0]):
        for j in range(np.shape(z_war)[1]):
            if z_cp[i,j] < 0:
                if -1*topo.z[7710+i,6390+j]>200:
                    thickness_cp[i,j] = 200
                else:
                    thickness_cp[i,j] = -1*topo.z[7710+i,6390+j]
            else:
                if z_cp[i,j] < 200:
                    thickness_cp[i,j] = z_cp[i,j]
                if z_cp[i,j] > 200:
                    if 200 - Z_cp[i,j] < 0:
                        thickness_cp[i,j] = 0
                    if 200 - Z_cp[i,j] > 0:
                        thickness_cp[i,j] = 200 - Z_cp[i,j]
                        
    c = plt.contourf(thickness_cp, levels = np.arange(0,210,10))
    #plt.scatter(101,76)
    plt.colorbar(c)
    plt.title('Cold Pool Thickness', fontsize = 14)
    plt.show()
    
    # thickness in terms of volume per little cube
    
    latitude = topo.y[7710:7862]
    longitude = topo.x[6390:6515]
    vol_thickness_cp = np.empty(np.shape(thickness_cp))
    vol_thickness_cp[:,:] = np.nan
    for i in range(np.shape(z_war)[0]):
        for j in range(np.shape(z_war)[1]):
            vol_thickness_cp[i,j] = thickness_cp[i,j]*0.001*((np.cos(np.radians(latitude[j]))*111.321)*(0.01666667))*((111.200*0.01666667))
    
    volume_cp = np.nansum(vol_thickness_cp) # total volume is the sum of all of these little cubes in km3

                        
    print('Volume of water below ' + str(omega_val) + ' = ' + str((volume_war/total_volume)*100))
    print('Volume of water in Cold Pool = ' + str((volume_cp/total_volume)*100))
    return thickness_cp, thickness_war
```

```python
#account for places with topographical mismatch.  
#if the value is negative just put 0 as the value (no thickness) and if the value is greater than 200 we will put 200 and not consider the deeper portion
thickness_t = np.empty([152,125])
thickness_t[:,:] = np.nan
for i in range(152):
    for j in range(125):
        if z_t[i,j] < 0:
            thickness_t[i,j] = -1*topo.z[7710+i,6390+j]
        if z_t[i,j] >= 0:
            if z_t[i,j] <= 200:
                thickness_t[i,j] = z_t[i,j]
            if z_t[i,j] > 200:
                if 200 - Z_t[i,j] < 0:
                    thickness_t[i,j] = 0
                if 200 - Z_t[i,j] >= 0:
                    thickness_t[i,j] = 200 - Z_t[i,j]
```

```python
#account for places with topographical mismatch.  
#if the value is negative just put 0 as the value (no thickness) and if the value is greater than 200 we will put 200 and not consider the deeper portion
thickness_s = np.empty([152,125])
thickness_s[:,:] = np.nan
for i in range(152):
    for j in range(125):
        if z_s[i,j] < 0:
            thickness_s[i,j] = -1*topo.z[7710+i,6390+j]
        if z_s[i,j] >= 0:
            if z_s[i,j] <= 200:
                thickness_s[i,j] = z_s[i,j]
            if z_s[i,j] > 200:
                if 200 - Z_s[i,j] < 0:
                    thickness_s[i,j] = 0
                if 200 - Z_s[i,j] >= 0:
                    thickness_s[i,j] = 200 - Z_s[i,j]
```
