```python
# Load NYB shapefile
NYB = gpd.read_file('~/Desktop/NYB_Indicators_Calculations/Datasets/Shapefiles/PlanningArea_NYocean_NYSDOS.shp')
# Alter the projection to WGS84 see https://epsg.io/4326
NYB = NYB.to_crs(epsg=4326)
```

```python
import regionmask

def crop_nd(data, longitude_name, latitude_name, shape):
    
    # Get the region of interest
    region = regionmask.from_geopandas(shape)
    
    # Create the mask
    mask = region.mask(data[longitude_name].astype('f4'), data[latitude_name].astype('f4'))
    
    # Apply mask to the data
    masked_ds = data.where(mask == region.numbers[0])
    
    return masked_ds
```

```python
gl_bt_1993 = xr.open_dataset('/Users/lauragruenburg/Desktop/NYB_Indicators_Calculations/CalculateIndicators/WaterTemperature/data/glorysBT_1993_2021.nc')
```

```python
gl_bt_1993_nyb = crop_nd(gl_bt_1993, 'longitude','latitude', NYB)
```

```python
gl_bt_1993df = gl_bt_1993_nyb.to_dataframe(['time','latitude','longitude'])
```

```python
gl_bt_1993df = gl_bt_1993df.reset_index()
```

```python
gl_bt_1993dfnyb = gl_bt_1993df[np.isnan(gl_bt_1993df.bottomT) == False]
```

```python
gl_bt_1993dfnyb.reset_index().to_csv('gl_bt_1993_2021nyb.csv')
```
