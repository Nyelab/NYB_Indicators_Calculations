```python
calfin_9 = calfin.loc[10:,:].reset_index(drop = True)
ctyp_9 = ctyp.loc[10:,:].reset_index(drop = True)
```

```python
calfin_9['stn_frac'] = calfin_9.station.astype('float') - np.floor(calfin_9.station.astype('float'))
ctyp_9['stn_frac'] = ctyp_9.station.astype('float') - np.floor(ctyp_9.station.astype('float'))
```

```python
calfin_9['shelf_dist'] = 'midshelf'
ctyp_9['shelf_dist'] = 'midshelf'
```

```python
calfin_9.loc[calfin_9.stn_frac < 0.3, 'shelf_dist'] = 'inshore'
calfin_9.loc[calfin_9.stn_frac > 0.4, 'shelf_dist'] = 'shelfbreak'
```

```python
ctyp_9.loc[ctyp_9.stn_frac < 0.3, 'shelf_dist'] = 'inshore'
ctyp_9.loc[ctyp_9.stn_frac > 0.4, 'shelf_dist'] = 'shelfbreak'
```

```python
fig, ax = plt.subplots(nrows = 2, ncols = 2, figsize = [10,10])
seaborn.boxplot(data = calfin_9[calfin_9.season == 'winter'], x = 'year', y = 'sumDensity', hue = 'shelf_dist', ax = ax[0,0])
ax[0,0].set_title('Winter')

seaborn.boxplot(data = calfin_9[calfin_9.season == 'spring'], x = 'year', y = 'sumDensity', hue = 'shelf_dist', ax = ax[0,1])
ax[0,1].set_title('Spring')

seaborn.boxplot(data = calfin_9[calfin_9.season == 'summer'], x = 'year', y = 'sumDensity', hue = 'shelf_dist', ax = ax[1,0])
ax[1,0].set_title('Summer')

seaborn.boxplot(data = calfin_9[calfin_9.season == 'autumn'], x = 'year', y = 'sumDensity', hue = 'shelf_dist', ax = ax[1,1])
ax[1,1].set_title('Autumn')

plt.show()
```

```python
fig, ax = plt.subplots(nrows = 2, ncols = 2, figsize = [10,10])
seaborn.boxplot(data = ctyp_9[ctyp_9.season == 'winter'], x = 'year', y = 'totalDens', hue = 'shelf_dist', ax = ax[0,0])
ax[0,0].set_title('Winter')

seaborn.boxplot(data = ctyp_9[ctyp_9.season == 'spring'], x = 'year', y = 'totalDens', hue = 'shelf_dist', ax = ax[0,1])
ax[0,1].set_title('Spring')

seaborn.boxplot(data = ctyp_9[ctyp_9.season == 'summer'], x = 'year', y = 'totalDens', hue = 'shelf_dist', ax = ax[1,0])
ax[1,0].set_title('Summer')

seaborn.boxplot(data = ctyp_9[ctyp_9.season == 'autumn'], x = 'year', y = 'totalDens', hue = 'shelf_dist', ax = ax[1,1])
ax[1,1].set_title('Autumn')

plt.show()
```
