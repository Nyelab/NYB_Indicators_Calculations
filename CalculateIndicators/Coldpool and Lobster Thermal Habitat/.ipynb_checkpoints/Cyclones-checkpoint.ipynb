```python
# import required module
import os
# assign directory
directory = '/Users/nyelab/Downloads/Cyclones/'
 
# iterate over files in
# that directory
for year_file in os.listdir(directory):
    if year_file != '.DS_Store':
        k=1
        for month_file in os.listdir(directory + year_file):
            if month_file != '.DS_Store':
                for file in os.listdir(directory + year_file + '/' + month_file):
                    f = os.path.join(directory, year_file, month_file, file)
                    # checking if it is a file
                    if os.path.isfile(f):
                        table = pn.read_csv(f)
                        step1 = table[table.lon>=-77]
                        step2 = step1[step1.lon<=-67]
                        step3 = step2[step2.lat <= 43]
                        final = step3[step3.lat >= 37]
                        if np.any(final) == True:
                            final['cyclone_id'] = k
                            cyclones = cyclones.append(final)
                            k=k+1
cyclones = cyclones.reset_index(drop = True)
cyclones = cyclones[cyclones.index > 0]                        
```

```python

cyclones['date'] = pn.to_datetime({'Year': cyclones.year,
                           'Month': cyclones.month,
                           'Day':cyclones.day,
                            'Hour': cyclones.hour})
cyclones['dayofyear'] = cyclones.date.dt.dayofyear
yearfrac = cyclones.dayofyear/365
cyclones['decimalyear'] = yearfrac + cyclones.year
cyclones2 = cyclones.sort_values(by=['date'])
cyclones2 = cyclones2.reset_index(drop=True)
```

```python
year = np.empty(len(cyclones3.groupby(['year','cyclone_id']).mean().strength[cyclones3.groupby(['year','cyclone_id']).mean().strength>1]))
num = np.empty(73)
num[0] = 0
a = 0
for i in range(len(cyclones3.groupby(['year','cyclone_id']).mean().strength[cyclones3.groupby(['year','cyclone_id']).mean().strength>1])):
    year[i] = cyclones3.groupby(['year','cyclone_id']).mean().strength[cyclones3.groupby(['year','cyclone_id']).mean().strength>1].index[i][0]
    if i == len(cyclones3.groupby(['year','cyclone_id']).mean().strength[cyclones3.groupby(['year','cyclone_id']).mean().strength>1])-1:
        num[a]=num[a]+1
        break
    if year[i] == cyclones3.groupby(['year','cyclone_id']).mean().strength[cyclones3.groupby(['year','cyclone_id']).mean().strength>1].index[i+1][0]:
        num[a]=num[a]+1
    if year[i] != cyclones3.groupby(['year','cyclone_id']).mean().strength[cyclones3.groupby(['year','cyclone_id']).mean().strength>1].index[i+1][0]:
        num[a] = num[a]+1
        a = a+1
    
```

```python
plt.plot(np.arange(1950,2023), num)
plt.tick_params(labelsize = 12)
plt.ylabel('Number of Large Cyclones', fontsize = 14)
```
