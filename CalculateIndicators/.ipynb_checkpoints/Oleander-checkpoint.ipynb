```python
salt = np.zeros([35,5,12])
```

```python
def salt_average(matrix, year, salt_mat):
    n = np.abs(1977 - year)
    bins = [73,73.2,73.4,73.6,73.8,74]
    months = np.unique(matrix['DATE'][:,1])
    for j in months:
        index = np.where(matrix['DATE'][:,1] == j)
        for i in range(len(bins)-1):
            ind = np.where(matrix['LON'][index]>= bins[i])
            ind2 = np.where(matrix['LON'][index]< bins[i+1])
            salt_mat[n,i,j-1] = np.nanmean(matrix['SfcSal'][index][np.intersect1d(ind[0], ind2[0])])
    return salt_mat
```
