```python

# Adapted from CoastWatch Python tutorial https://github.com/coastwatch-training/CoastWatch-Tutorials/blob/main/extract-satellite-data-within-boundary/Python/extract-satellite-data-within-boundary.ipynb

# Given an xarray n dimensional dataset and a geopandas geodataframe, 
# the xarray dataset will mask all values outside of the desired shape with nans

# Note this only works for a single shape.
# If your geodataframe has multiple polygons you can loop through each one.

# data is an N dimensional xarray dataset
# longitude name is a string of the variable name for longitude in data
# latitude name is a string of the variable name for latitude in data
# shape is a geodataframe

def crop_nd(data, longitude_name, latitude_name, shape):
    
    # Get the region of interest
    region = regionmask.from_geopandas(shape)
    
    # Create the mask
    mask = region.mask(data[longitude_name].astype('f4'), data[latitude_name].astype('f4'))
    
    # Apply mask to the data
    masked_ds = data.where(mask == region.numbers[0])
    
    return masked_ds
```

```python
#10-21
data['geometry'] = gpd.points_from_xy(data.lon, data.lat, crs="EPSG:4269")
data_gpd = gpd.GeoDataFrame(data=data, geometry='geometry')

data_strata = gpd.sjoin(left_df=data_gpd, right_df=Strata[Strata['STR_NAME'] == 10], how='left')
data_strata = data_strata[data_strata['TYPE'] == Strata[Strata['STR_NAME'] == 14].TYPE[14-1]]

for i in range(11, 22):
    data_new = gpd.sjoin(left_df=data_gpd, right_df=Strata[Strata['STR_NAME'] == i], how='left')
    data_new = data_new[data_new['TYPE'] == Strata[Strata['STR_NAME'] == i].TYPE[i-1]]
    data_strata = pd.concat([data_strata, data_new])
```

```python
data_strata = data_strata[data_strata.bot_temp >=0.1]
data_strata = data_strata[data_strata.bot_temp <=30]
```

```python
Coastal = data_strata[data_strata['TYPE'] == 'C'].reset_index(drop = True)
Inshore = data_strata[data_strata['TYPE'] == 'IS'].reset_index(drop = True)
Midshelf = data_strata[data_strata['TYPE'] == 'MS'].reset_index(drop = True)
Shelfbreak = data_strata[data_strata['TYPE'] == 'SB'].reset_index(drop = True)
```

```python
C_botT = Coastal.groupby(pd.PeriodIndex(Coastal['date'], freq='M'))['bot_temp'].mean()
I_botT = Inshore.groupby(pd.PeriodIndex(Inshore['date'], freq='M'))['bot_temp'].mean()
M_botT = Midshelf.groupby(pd.PeriodIndex(Midshelf['date'], freq='M'))['bot_temp'].mean()
S_botT = Shelfbreak.groupby(pd.PeriodIndex(Shelfbreak['date'], freq='M'))['bot_temp'].mean()
```

```python
CT = pd.DataFrame({'time': C_botT.index.astype('datetime64[ns]'), 
                   'temp': C_botT.values})

IT = pd.DataFrame({'time': I_botT.index.astype('datetime64[ns]'), 
                   'temp': I_botT.values})

MT = pd.DataFrame({'time': M_botT.index.astype('datetime64[ns]'), 
                   'temp': M_botT.values})

ST = pd.DataFrame({'time': S_botT.index.astype('datetime64[ns]'), 
                   'temp': S_botT.values})
```

```python
def get_season(date):
    month = date.month
    if 3 <= month <= 5:
        return 'Spring'
    elif 6 <= month <= 8:
        return 'Summer'
    elif 9 <= month <= 11:
        return 'Autumn'
    else:
        return 'Winter'

# Apply the function to create a new 'season' column
CT['season'] = CT['time'].apply(get_season)
IT['season'] = IT['time'].apply(get_season)
MT['season'] = MT['time'].apply(get_season)
ST['season'] = ST['time'].apply(get_season)
```

```python
CT_seas = CT.groupby(['season', CT['time'].dt.year]).mean('season').reset_index()
IT_seas = IT.groupby(['season', IT['time'].dt.year]).mean('season').reset_index()
MT_seas = MT.groupby(['season', MT['time'].dt.year]).mean('season').reset_index()
ST_seas = ST.groupby(['season', ST['time'].dt.year]).mean('season').reset_index()
```

```python
fig, ax = plt.subplots(nrows = 2, ncols = 2, figsize = [10,10])
ax[0,0].plot(CT_seas[CT_seas.season == 'Winter'].time,
         CT_seas[CT_seas.season == 'Winter'].temp, c = 'cyan', alpha = 0.5)
ax[0,0].plot(IT_seas[IT_seas.season == 'Winter'].time,
         IT_seas[IT_seas.season == 'Winter'].temp, c = 'blue', alpha = 0.5)
ax[0,0].plot(MT_seas[MT_seas.season == 'Winter'].time,
         MT_seas[MT_seas.season == 'Winter'].temp, c = 'green', alpha = 0.5)
ax[0,0].plot(ST_seas[ST_seas.season == 'Winter'].time,
         ST_seas[ST_seas.season == 'Winter'].temp, c = 'purple', alpha = 0.5)
ax[0,0].set_title('DJF')

ax[0,1].plot(CT_seas[CT_seas.season == 'Spring'].time,
         CT_seas[CT_seas.season == 'Spring'].temp, c = 'cyan', alpha = 0.5)
ax[0,1].plot(IT_seas[IT_seas.season == 'Spring'].time,
         IT_seas[IT_seas.season == 'Spring'].temp, c = 'blue', alpha = 0.5)
ax[0,1].plot(MT_seas[MT_seas.season == 'Spring'].time,
         MT_seas[MT_seas.season == 'Spring'].temp, c = 'green', alpha = 0.5)
ax[0,1].plot(ST_seas[ST_seas.season == 'Spring'].time,
         ST_seas[ST_seas.season == 'Spring'].temp, c = 'purple', alpha = 0.5)
ax[0,1].set_title('MAM')

ax[1,0].plot(CT_seas[CT_seas.season == 'Summer'].time,
         CT_seas[CT_seas.season == 'Summer'].temp, c = 'cyan', alpha = 0.5)
ax[1,0].plot(IT_seas[IT_seas.season == 'Summer'].time,
         IT_seas[IT_seas.season == 'Summer'].temp, c = 'blue', alpha = 0.5)
ax[1,0].plot(MT_seas[MT_seas.season == 'Summer'].time,
         MT_seas[MT_seas.season == 'Summer'].temp, c = 'green', alpha = 0.5)
ax[1,0].plot(ST_seas[ST_seas.season == 'Summer'].time,
         ST_seas[ST_seas.season == 'Summer'].temp, c = 'purple', alpha = 0.5)
ax[1,0].set_title('JJA')

ax[1,1].plot(CT_seas[CT_seas.season == 'Autumn'].time,
         CT_seas[CT_seas.season == 'Autumn'].temp, c = 'cyan', alpha = 0.5, label = 'C')
ax[1,1].plot(IT_seas[IT_seas.season == 'Autumn'].time,
         IT_seas[IT_seas.season == 'Autumn'].temp, c = 'blue', alpha = 0.5, label = 'IS')
ax[1,1].plot(MT_seas[MT_seas.season == 'Autumn'].time,
         MT_seas[MT_seas.season == 'Autumn'].temp, c = 'green', alpha = 0.5, label = 'MS')
ax[1,1].plot(ST_seas[ST_seas.season == 'Autumn'].time,
         ST_seas[ST_seas.season == 'Autumn'].temp, c = 'purple', alpha = 0.5, label = 'SB')
ax[1,1].set_title('SON')
ax[1,1].legend()
plt.show()
```
